```python
from pathlib import Path

project_dir = Path.cwd()

print("Current working directory:", project_dir)
print("Project directory exists:", project_dir.exists())

csv_files = [
file.relative_to(project_dir)
for file in project_dir.rglob("*.csv")
if ".venv" not in file.parts and ".git" not in file.parts
]

print("\nCSV files found:")
print(*csv_files, sep="\n")
```

```python
from pathlib import Path

project_dir = Path.cwd().parent
extensions = {".csv", ".xlsx", ".xls", ".sav", ".dta", ".parquet"}

data_files = [
str(file.relative_to(project_dir))
for file in project_dir.rglob("*")
if file.is_file()
and file.suffix.lower() in extensions
and not any(
part.lower() in {".venv", ".venv313", ".git", "node_modules"}
for part in file.parts
)
]

print("Project directory:", project_dir)
print("\nResearch data files found:")
print("\n".join(data_files) if data_files else "No matching files found.")
```
